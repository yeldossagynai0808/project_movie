```python
# =========================================================
# Задача 1 — Загрузка и первичный просмотр данных
# =========================================================
print("=== Задача 1 ===")
print("Первые 5 строк:")
print(df.head())
print(f"\nФорма DataFrame (строк, колонок): {df.shape}")
print("\nТипы данных всех колонок:")
print(df.dtypes)
print("\nКоличество пропусков в каждой колонке:")
print(df.isnull().sum())
print("-" * 50)
```

```python
# =========================================================
# Задача 2 — Уникальные значения и списки
# =========================================================
print("=== Задача 2 ===")
types_list = df["Type"].unique().tolist()
genres_list = df["Genre"].unique().tolist()

print("types_list =", types_list)
print("genres_list =", genres_list)
print("-" * 50)
```

```python
# =========================================================
# Задача 3 — Работа с названиями фильмов
# =========================================================
print("=== Задача 3 ===")
# Создаем список названий в верхнем регистре
upper_titles = df["Title"].str.upper().tolist()

# Создаем список длин каждого названия (убрали внешний len)
title_lengths = df["Title"].str.len().tolist()

# Считаем среднюю длину
mean_length = df["Title"].str.len().mean()

print(f"Первые 5 названий в верхнем регистре: {upper_titles[:5]}")
print(f"Первые 5 длин названий: {title_lengths[:5]}")
print(f"Средняя длина названия фильма: {mean_length:.2f}")
print("-" * 50)
```

```python
# =========================================================
# Задача 4 — Фильтрация фильмов по рейтингу
# =========================================================
print("=== Задача 4 ===")
top_rated = df[df['Rating_IMDB'] >= 8.5]
print("Топ-10 фильмов по рейтингу (>= 8.5):")
print(top_rated[["Title", "Rating_IMDB", "Genre"]].head(10))
print("-" * 50)
```

```python
# =========================================================
# Задача 5 — Топ по доходу
# =========================================================
print("=== Задача 5 ===")
df["Profit"] = df["Revenue_million"] - df["Budget_million"]
top_profit = df.sort_values(by="Profit", ascending=False)

print("Топ-10 самых прибыльных фильмов:")
print(top_profit[["Title", "Profit", "Rating_IMDB"]].head(10))
print("-" * 50)
```

```python
# =========================================================
# Задача 6 — Сохранение промежуточных результатов
# =========================================================
print("=== Задача 6 ===")
# Сохраняем отфильтрованный ранее top_rated в Excel без индексов
top_rated.to_excel("student1_top_rated.xlsx", index=False)
print("Файл 'student1_top_rated.xlsx' успешно сохранен.")
print("-" * 50)
```

```python
# =========================================================
# Задача 7 — Поиск по ключевому слову
# =========================================================
print("=== Задача 7 ===")
# Ищем фильмы, где в Title содержится "Movie_1" (регистронезависимо на всякий случай)
keyword_search = df[df["Title"].str.contains("Movie_1", na=False)]
print("Результаты поиска по ключевому слову 'Movie_1':")
print(keyword_search[["Title", "Genre", "Rating_IMDB"]])
print("-" * 50)
```

```python
# =========================================================
# Задача 8 — Сортировка фильмов
# =========================================================
print("=== Задача 8 ===")
# 1. Сортировка по количеству голосов (по убыванию)
top_votes = df.sort_values(by="Votes_IMDB", ascending=False).head(10)
print("Топ-10 фильмов с наибольшим количеством голосов:")
print(top_votes[["Title", "Votes_IMDB"]])

print("\n2. Сортировка по длительности (по возрастанию):")
# 2. Сортировка по длительности (по возрастанию)
shortest_movies = df.sort_values(by="Duration_min", ascending=True).head(10)
print(shortest_movies[["Title", "Duration_min"]])
print("-" * 50)
```

```python
# =========================================================
# Задача 9 — Создание нового признака
# =========================================================
print("=== Задача 9 ===")
# Считаем стоимость на один голос зрителя
df["budget_per_vote"] = df["Budget_million"] / df["Votes_IMDB"]

# Находим топ-5 с минимальным значением
top_efficient_movies = df.sort_values(by="budget_per_vote", ascending=True).head(5)
print("Топ-5 фильмов с минимальным бюджетом на один голос:")
print(top_efficient_movies[["Title", "Budget_million", "Votes_IMDB", "budget_per_vote"]])
print("-" * 50)
```

```python
# Задача 10 — Генератор фильмов по жанру
# =========================================================
print("=== Задача 10 ===")

# Функция-генератор, которая лениво читает строки нужного жанра
def action_movies_generator(dataframe):
    # Фильтруем строки, где жанр равен 'Action'
    action_df = dataframe[dataframe["Genre"] == "Action"]
    for index, row in action_df.iterrows():
        yield row["Title"], row["Genre"], row["Rating_IMDB"]

# Создаем экземпляр генератора
gen = action_movies_generator(df)

print("Первые 20 фильмов жанра 'Action' из генератора:")
for i in range(20):
    try:
        movie_info = next(gen)
        print(f"{i+1}. Title: {movie_info[0]} | Genre: {movie_info[1]} | Rating: {movie_info[2]}")
    except StopIteration:
        print("Фильмы в генераторе закончились.")
        break
print("-" * 50)
```
